### Exercise 1: Relative and Global Phase

In this exercise, we will explore the difference between the relative
phase and the global phase of a quantum state.

A **global phase** multiplies all amplitudes by the same phase factor.
For example, $\ket{\Psi}$ and $-\ket{\Psi}$ differ only by an overall
minus sign. This does not change any measurement probabilities.

A **relative phase** describes the phase difference between amplitudes.
For example,
$\ket{\Psi_1} = \frac{1}{\sqrt{2}}(\ket{0}+\ket{1})$ and
$\ket{\Psi_2} = \frac{1}{\sqrt{2}}(\ket{0}-\ket{1})$
differ in the sign of one amplitude.
We can distinguish these states by applying suitable gates and then measuring.

**Observation:** All three states give $p(0)=p(1)=1/2$.

do not mean that the states are the same: $\ket{\Psi_1}$ and
$\ket{\Psi_2}$ have different **relative phases**. In contrast,
$\ket{\Psi_3}=-\ket{\Psi_2}$: these two states differ only by a
**global phase**. We will explore this distinction in the next exercise.


**Exercise 1.5:** Apply one more Hadamard gate to each of the three
states. Then display the resulting states and their measurement probabilities.

Which states can now be distinguished by a measurement?
In particular, explain the difference between $\ket{\Psi_2}$ and
$\ket{\Psi_3}$: could applying the same further sequence of gates to both
states make their measurement probabilities different?


```python
quantum_circuit_1.h(0)
quantum_circuit_2.h(0)
quantum_circuit_3.h(0)

display(sv_state(quantum_circuit_1, "State 1 after H: "))
display(sv_state(quantum_circuit_2, "State 2 after H: "))
display(sv_state(quantum_circuit_3, "State 3 after H: "))
display(sv_probs(quantum_circuit_1, "Probabilities after H, state 1: "))
display(sv_probs(quantum_circuit_2, "Probabilities after H, state 2: "))
display(sv_probs(quantum_circuit_3, "Probabilities after H, state 3: "))
```

**Observation:** We obtain

$$H\ket{\Psi_1}=\ket{0},\qquad
H\ket{\Psi_2}=\ket{1},\qquad
H\ket{\Psi_3}=-\ket{1}.$$

After the Hadamard gate, state 1 always gives the measurement outcome 0,
whereas states 2 and 3 always give the outcome 1.
The difference in **relative phase** between states 1 and 2 has become
visible in their measurement probabilities.

States 2 and 3, however, differ only by an overall factor of $-1$.
This is a **global phase**. They represent the same physical state and
remain indistinguishable after any further gate sequence $U$ applied to both:

$$U(-\ket{\Psi_2})=-U\ket{\Psi_2}.$$

The factor $-1$ disappears when we take the squared magnitude of each amplitude.

### Exercise 3: Measurements

So far, we have used a simulator to calculate quantum states and their
theoretical measurement probabilities. A single measurement outcome
does not tell us these probabilities. We estimate them by running the
same circuit many times, preparing the state and measuring it on each run.

In Qiskit, **classical bits** store the measurement outcomes.
In `QuantumCircuit(num_qubits, num_bits)`, the first argument specifies
the number of qubits and the second the number of classical bits.

The instruction `quantum_circuit.measure(0, 0)` measures qubit 0 and stores
the outcome in classical bit 0. The first index identifies the qubit;
the second identifies the classical bit.


```python
# One qubit and one classical bit
quantum_circuit = QuantumCircuit(1, 1)
quantum_circuit.h(0)
quantum_circuit.measure(0, 0)
quantum_circuit.draw("mpl")
```

In the circuit diagram, `c` denotes the classical register that
stores the measurement outcomes. The double line represents classical
information: the stored values are 0 or 1.

#### Simulating Measurements with Qiskit

A **shot** is one execution of a circuit, including its measurements.
For each shot, the qubits are prepared in the initial state again and
all gates are applied again. Each shot therefore uses a freshly prepared
state; it does not repeatedly measure a state that has already been measured.

The helper function `sample_from_circuit` performs these measurements
on a simulator. For example, use
`sample_from_circuit(quantum_circuit, num_shots=1024)` to run 1024 shots.
The argument `num_shots` must be a positive integer.

The result is a dictionary containing the **counts** of the measured
bitstrings. For example, `{"00": 15, "01": 20, "10": 35, "11": 30}`
means that, out of 100 shots in total, the bitstring `"00"` was measured
15 times. A bitstring that was never measured may be absent from the dictionary.


```python
from qc_lecture_tools.sampling import sample_from_circuit

# Example: only the first qubit is placed in a superposition.
quantum_circuit = QuantumCircuit(2, 2)
quantum_circuit.h(0)

display(sv_state(quantum_circuit, "Quantum state before measurement: "))
display(sv_probs(quantum_circuit, "Theoretical measurement probabilities: "))

quantum_circuit.measure(0, 0)
quantum_circuit.measure(1, 1)

sample_from_circuit(quantum_circuit, num_shots=1000)
```

**Observation:** The second qubit remains in $\ket{0}$. Only the
bitstrings `00` and `10` can occur, each with a theoretical probability
of $1/2$. With 1000 shots, their counts will usually be close to 500 each,
but they need not be exactly equal. The helper function returns bitstrings
in the same order used throughout this notebook: the first qubit comes first.

**Exercise 3.3:** Convert the counts in `measurements` into
**relative frequencies**. These provide estimates of the measurement probabilities:

$$p_{\mathrm{estimated}}(\text{bitstring})
=\frac{\text{number of times this bitstring was measured}}{\text{total number of shots}}.$$

Write a function `measure_to_probability(measurements)` that returns
a dictionary of estimated probabilities.
You can calculate the total number of shots using `sum(measurements.values())`.

**Reminder:** Here is how to access the contents of a Python dictionary:


```python
results = {"00": 15, "01": 20, "10": 35, "11": 30}

print("Number of times 00 was measured:", results["00"])
print("Measured bitstrings:", results.keys())
print("Counts:", results.values())

for state in results.keys():
    print(f"The bitstring {state} was measured {results[state]} times.")
```

```python
def measure_to_probability(measurements):
    probs = {}
    total_counts = sum(measurements.values())
    for state in measurements.keys():
        probs[state] = measurements[state] / total_counts
    return probs
```

```python
# Reuse the results of the 500 shots from Exercise 3.2
probabilities = measure_to_probability(measurements)
print(probabilities)
print("Sum of the estimated probabilities:", sum(probabilities.values()))
```

**Observation:** Each count is divided by the total number of shots.
The estimated probabilities sum to 1, apart from possible rounding errors,
and are approximately $0.25$ each in this example.
